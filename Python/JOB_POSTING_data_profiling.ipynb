```python
#as we can see bengaluru, then group of other cities and bengaluru too,plus "hybrid"
#so we gonna remove hybrid , #then seperate locations from comma seperated location

def clean_location(loc):
    loc = str(loc)
    
    # Remove "Hybrid - " prefix if present
    loc = loc.replace('Hybrid - ', '').strip()
    
    # If multiple cities are listed (comma-separated), just take the first one
    primary_city = loc.split(',')[0].strip()
    
    return primary_city

data['primary_location'] = data['location'].apply(clean_location)

# Separate flag: was this job listed as hybrid?
data['is_hybrid'] = data['location'].str.contains('Hybrid', case=False, na=False)

print(data['primary_location'].value_counts().head(20))
print("\nHybrid jobs:", data['is_hybrid'].sum())
```

```python
#now we have minimum and max experience col in data set , we give the category to numbers, instead of flot no



# Rename to match our project's naming convention (lowercase, consistent style)
data['min_experience'] = data['minimumExperience']
data['max_experience'] = data['maximumExperience']

def experience_group(min_exp):
    if pd.isnull(min_exp):
        return 'Not specified'
    elif min_exp == 0:
        return 'Fresher / 0-1'
    elif min_exp <= 3:
        return '1-3 years'
    elif min_exp <= 5:
        return '3-5 years'
    elif min_exp <= 8:
        return '5-8 years'
    else:
        return '8+ years'

data['experience_group'] = data['min_experience'].apply(experience_group)

print(data['experience_group'].value_counts())
```

```python
#In salary , we got mini /max columns , some contains 0 as salary , we can define them as not disclosed 
# Rename for consistency
data['min_salary'] = data['minimumSalary']
data['max_salary'] = data['maximumSalary']

# Treat 0 as "not disclosed" -> convert to missing (NaN) instead of a fake real salary
data.loc[data['min_salary'] == 0, 'min_salary'] = None
data.loc[data['max_salary'] == 0, 'max_salary'] = None

# Average salary, only where both min and max are actually known
data['avg_salary'] = (data['min_salary'] + data['max_salary']) / 2

print(data[['min_salary', 'max_salary', 'avg_salary']].describe())
print("\nJobs with disclosed salary:", data['avg_salary'].notnull().sum())
print("Jobs with salary NOT disclosed:", data['avg_salary'].isnull().sum())
```

```python
import re
from datetime import datetime, timedelta

REFERENCE_DATE = datetime(2025, 10, 15)  # dataset's approximate scrape date

def parse_posting_date(text):
    text = str(text).strip()
    
    if text == 'Today':
        return REFERENCE_DATE
    if text == 'Just Now':
        return REFERENCE_DATE
    
    match = re.match(r'(\d+)\s+Days?\s+Ago', text, re.IGNORECASE)
    if match:
        days = int(match.group(1))
        return REFERENCE_DATE - timedelta(days=days)
    
    if 'Few Hours Ago' in text:
        return REFERENCE_DATE
    
    # Anything else ("Starts...", "Starts in...") -> not a real posting date
    return None

data['posting_date'] = data['jobUploaded'].apply(parse_posting_date)
data['is_future_start'] = data['jobUploaded'].str.contains('Starts', case=False, na=False)

# Break posting_date into useful parts for trend analysis
data['posting_year'] = data['posting_date'].dt.year
data['posting_month'] = data['posting_date'].dt.month
data['posting_week'] = data['posting_date'].dt.isocalendar().week

print("Jobs with valid posting_date:", data['posting_date'].notnull().sum())
print("Jobs flagged as future start:", data['is_future_start'].sum())
print(data['posting_date'].describe())
```

```python
def split_skills(tags):
    if pd.isnull(tags):
        return []
    # Split on comma, strip whitespace from each skill
    return [skill.strip() for skill in str(tags).split(',') if skill.strip()]

data['skills_list'] = data['tagsAndSkills'].apply(split_skills)

print(data['skills_list'].head(5))
print("\nJobs with at least one skill:", (data['skills_list'].str.len() > 0).sum())
```

```python
# "Explode" turns each list into multiple rows - one row per skill, same job_id repeated
job_skills = data[['jobId', 'skills_list']].explode('skills_list')

# Rename for clarity
job_skills = job_skills.rename(columns={'skills_list': 'skill'})

# Drop any empty/missing skill rows (from jobs that had no tagsAndSkills)
job_skills = job_skills.dropna(subset=['skill'])

print(job_skills.head(10))
print("\nTotal job-skill rows:", job_skills.shape[0])
print("Unique skills (raw, unstandardized):", job_skills['skill'].nunique())
```

```python
job_skills['skill'].unique()
```

```python
def standardize_skill(skill):
    skill_lower = str(skill).lower().strip()
    
    # Data / Analytics
    if 'sql' in skill_lower:
        return 'SQL'
    elif 'python' in skill_lower:
        return 'Python'
    elif 'excel' in skill_lower:
        return 'Excel'
    elif 'power bi' in skill_lower or 'powerbi' in skill_lower:
        return 'Power BI'
    elif 'tableau' in skill_lower:
        return 'Tableau'
    elif 'pandas' in skill_lower:
        return 'Pandas'
    elif 'numpy' in skill_lower:
        return 'NumPy'
    elif 'statistics' in skill_lower or 'statistical' in skill_lower:
        return 'Statistics'
    elif 'data visualization' in skill_lower or 'data viz' in skill_lower:
        return 'Data Visualization'
    elif 'bigquery' in skill_lower or 'big query' in skill_lower:
        return 'BigQuery'
    elif 'snowflake' in skill_lower:
        return 'Snowflake'
    
    # Data Engineering
    elif 'etl' in skill_lower:
        return 'ETL'
    elif 'airflow' in skill_lower:
        return 'Airflow'
    elif 'spark' in skill_lower:
        return 'Spark'
    elif 'databricks' in skill_lower:
        return 'Databricks'
    elif 'kafka' in skill_lower:
        return 'Kafka'
    
    # Cloud
    elif 'aws' in skill_lower or 'amazon web services' in skill_lower:
        return 'AWS'
    elif 'azure' in skill_lower:
        return 'Azure'
    elif 'gcp' in skill_lower or 'google cloud' in skill_lower:
        return 'GCP'
    
    # AI / ML
    elif 'machine learning' in skill_lower or skill_lower == 'ml':
        return 'Machine Learning'
    elif 'deep learning' in skill_lower:
        return 'Deep Learning'
    elif 'nlp' in skill_lower or 'natural language' in skill_lower:
        return 'NLP'
    elif 'llm' in skill_lower:
        return 'LLM'
    elif 'tensorflow' in skill_lower:
        return 'TensorFlow'
    elif 'pytorch' in skill_lower:
        return 'PyTorch'
    
    # Business / Product
    elif 'business analysis' in skill_lower:
        return 'Business Analysis'
    elif 'a/b testing' in skill_lower or 'ab testing' in skill_lower:
        return 'A/B Testing'
    elif 'kpi' in skill_lower:
        return 'KPI'
    
    else:
        return None  # not in our controlled dictionary - not a data/tech skill we're tracking

job_skills['standardized_skill'] = job_skills['skill'].apply(standardize_skill)

print(job_skills['standardized_skill'].value_counts())
print("\nRows matched to a known skill:", job_skills['standardized_skill'].notnull().sum())
print("Rows NOT matched (returned None):", job_skills['standardized_skill'].isnull().sum())
```

```python
# Keep only rows that matched our controlled skill dictionary
job_skills_clean = job_skills.dropna(subset=['standardized_skill']).copy()

# Keep only the columns we actually need going forward
job_skills_clean = job_skills_clean[['jobId', 'standardized_skill']]
job_skills_clean = job_skills_clean.rename(columns={'standardized_skill': 'skill'})

# Some jobs might list the same standardized skill twice (e.g. "Excel" and "Advanced Excel" both -> "Excel")
# Drop those duplicate job-skill pairs
job_skills_clean = job_skills_clean.drop_duplicates()

print(job_skills_clean.head(10))
print("\nFinal job_skills row count:", job_skills_clean.shape[0])
print("Jobs represented in this table:", job_skills_clean['jobId'].nunique())
```
